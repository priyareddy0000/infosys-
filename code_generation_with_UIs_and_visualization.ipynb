excecuting all the five codes in one code

```python
prompt = "Write a Python function to check if a number is prime."

def generate_code(prompt, tokenizer, model):
    inputs = tokenizer(prompt, return_tensors="pt")
    outputs = model.generate(inputs.input_ids, max_new_tokens=256)
    code = tokenizer.decode(outputs[0], skip_special_tokens=True)
    return code

print("=== DeepSeek-Coder ===")
print(generate_code(prompt, tokenizer_ds, model_ds))

print("=== Phi-2 ===")
print(generate_code(prompt, tokenizer_phi, model_phi))

print("=== Gemma ===")
print(generate_code(prompt, tokenizer_gemma, model_gemma))

print("=== Stable-Code ===")
print(generate_code(prompt, tokenizer_stable, model_stable))

print("=== Replit-Code ===")
print(generate_code(prompt, tokenizer_replit, model_replit))
```

1.Bubble Sort



```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr
```

2.Linked list Class

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None

    def insert(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
        else:
            temp = self.head
            while temp.next:
                temp = temp.next
            temp.next = new_node

    def display(self):
        temp = self.head
        while temp:
            print(temp.data, end=" ")
            temp = temp.next
        print()
```

3.Check prime

```python
def is_prime(num):
    if num <= 1:
        return False
    for i in range(2, int(num**0.5)+1):
        if num % i == 0:
            return False
    return True
```

5.Word Count in File

```python
def count_words(filename):
    with open(filename, 'r') as f:
        text = f.read()
    return len(text.split())
```
